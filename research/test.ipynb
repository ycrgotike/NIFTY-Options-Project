```python
import yfinance as yf
import pandas as pd

# NIFTY 50 Index
ticker = "^NSEI"

# Download 1 year of daily data
df = yf.download(ticker, period="1y", interval="1d")

print(df.head())
print(df.dtypes)
```

```python
df.columns = df.columns.get_level_values(0)
```

```python
print(df.columns)
```

```python
print(df.dtypes)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plt.plot(df["Close"])
plt.title("NIFTY Close Price")
plt.xlabel("Date")
plt.ylabel("Price")
plt.grid(True)
plt.show()
```

```python
df["MA20"] = df["Close"].rolling(window=20).mean()
```

```python
print(df[["Close", "MA20"]].tail())
```

```python
plt.figure(figsize=(12,6))

plt.plot(df.index, df["Close"], label="Close")
plt.plot(df.index, df["MA20"], label="20 MA")

plt.title("NIFTY with 20-Day Moving Average")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.grid(True)

plt.show()
```
